EDA Visualization

```python
#plt.hist() – Displays the distribution of students' marks by grouping them into 5 ranges (bins).
plt.hist(df["Marks"], bins=5)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Marks Distribution")

plt.show()
```

```python
#sns.countplot() – Shows the number of students in each department.
sns.countplot(data=df, x="Department")

plt.title("Students by Department")
plt.show()
```

```python
#sns.scatterplot() – Shows the relationship between Study Hours and Marks for each student.
sns.scatterplot(data=df, x="Study_Hours", y="Marks")

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
#sns.scatterplot() – Shows the relationship between Attendance and Marks for each student.
sns.scatterplot(data=df, x="Attendance", y="Marks")

plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.title("Attendance vs Marks")

plt.show()
```

```python
#Calculates the correlation between Study Hours, Attendance, and Marks to show how strongly they are related.
df[["Study_Hours", "Attendance", "Marks"]].corr()
```

```python
#sns.heatmap() – Visualizes the correlation between Study Hours, Attendance, and Marks using colors and correlation values.
sns.heatmap(
    df[["Study_Hours", "Attendance", "Marks"]].corr(),
    annot=True
)

plt.show()
```
